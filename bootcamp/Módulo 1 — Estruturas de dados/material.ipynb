### Uma lista e uma colecao ordenada e mutavel de objetos Python.

```python
mensagens = [
    "Meu pedido atrasou",
    "Quero cancelar minha compra",
    "Preciso de um reembolso"
]

print(mensagens[0]) # Primeiro elemento
print(mensagens[1]) # Segundo elemento
print(mensagens[2]) # Terceiro elemento
print(mensagens[-1]) # Ultimo elemento

print(len(mensagens)) # Quantidade de mensagens 

mensagens.append("Não consigo acessar minha conta")
print(mensagens)
```

Operação	Resultado
lista[0]	Primeiro elemento
lista[-1]	Último elemento
lista.append(x)	Adiciona um elemento ao final
lista.extend(xs)	Adiciona vários elementos
lista.pop()	Remove e retorna o último
lista.remove(x)	Remove a primeira ocorrência de x
len(lista)	Quantidade de elementos
x in lista	Verifica a presença de x
lista[1:4]	Obtém uma fatia da lista

```python
# .extend() -> Adiciona vários elementos

mensagens = [
    "Meu pedido atrasou",
    "Quero cancelar minha compra",
    "Preciso de um reembolso"
]


adicionar = [
    "Meu pedido veio errado",
    "Meu pedido não chegou",
    "O motoboy foi mal educado"
]

mensagens.extend(adicionar)

print(mensagens)
```

```python
# .pop() Remove e retorna o último ítem da lista

mensagens = [
    "Meu pedido atrasou",
    "Quero cancelar minha compra",
    "Preciso de um reembolso",
    "Meu pedido veio errado",
    "Meu pedido não chegou",
    "O motoboy foi mal educado"
]

removido = mensagens.pop()
print(removido)

```

```python
# .remove() -> Remove um item da lista 

lista = [1,2,3,4,5,6]
lista.remove(5)
print(lista)
```

```python
# x in lista -> Verifica a presença de x in lista

numeros = [10,20,30,40]

print(20 in numeros)

#############################################

produtos = ["Notebook", "Mouse", "Teclado"]

produto = "Mouse"

if produto in produtos:
    print("Produto encontrado")
else:
    print("Produto não encontrado")
```

```python
# lista[inicio:fim:passo]

# inicio: índice onde começa a extração (inclusivo).
# fim: índice onde termina a extração (exclusivo).
# passo: intervalo entre os elementos (opcional; padrão 1).

# lista[:3] -> 	Primeiros 3 elementos 
# lista[2:] ->  Do índice 2 até o final
# lista[:]  ->  Cópia superficial da lista
# lista[::2] -> Elementos de 2 em 2
# lista[-3:] -> Últimos 3 elementos
# lista[::-1] -> Lista em ordem inversa
# lista[1:5:2] -> Índices 1 e 3

lista = [10, 20, 30, 40, 50, 60]

print(lista[:]) # Traz toda a lista  print(lista[0:]) é a mesma coisa
print(lista[1:])  # Traz lista a partir do segundo indice
print(lista[4::3])

```

```python
# 5. Aplicação em Engenharia de Dados
# Imagine que você recebeu 1.000 registros de uma API e deseja processá-los em lotes de 100.
# range(inicio,fim,passo)

registros = list(range(1,1001))
tamanho_lote = 100

for inicio in range(0,len(registros),tamanho_lote):
    lote = registros[inicio:inicio + tamanho_lote]

    print(
        f"Processando {len(lote)} registros " f"do ID {lote[0]} ao {lote[-1]}"
        )
```

### 1.2 Tuplas — tuple


```python
# A diferença principal entre a tupla e a lista é que não podemos substituir diretamente os elementos da tupla.


registro = (101, "entrega", "pendente")

print(registro[0]) # Traz o primeiro item da tupla
print(registro[1]) # Traz o segundo item da tupla
print(registro[2]) # Traz o terceiro item da tupla
print(registro[-1]) # Traz o ultimo item da tupla
# registro[0] = 200 - TypeError
```

```python
#  Desempacotamento 
pedido_id, categoria, status = registro
print(pedido_id)
print(categoria)
print(status)
```

```python
# Isso é bastante utilizado quando funções retornam múltiplos valores.

def calcular_metricas(valores):
    total = sum(valores)
    media = total/len(valores) if valores else 0
    return total, media 

faturamento, ticket_medio = calcular_metricas(
    [100,200,300]
)


print(faturamento)
print(ticket_medio)
```

### 1.3 Conjuntos — set


```python
# Um set armazena elementos únicos e permite operações matemáticas de conjuntos.

categorias = [
    "entrega",
    "pagamento",
    "entrega",
    "cancelamento",
    "pagamento"
]

categorias_unicas = set(categorias)

print(categorias_unicas)

```

```python
# Operações de conjuntos
clientes_ativos = {101,102,103,104}
clientes_com_pedidos = {102,104,105}

# Interseção: aparecem nos dois
print(clientes_ativos & clientes_com_pedidos)

# União: aparecem em pelo menos um
print(clientes_ativos | clientes_com_pedidos)

# Diferença: ativos sem pedidos
print(clientes_ativos - clientes_com_pedidos)

# Diferença simétrica - Ele retorna os elementos que estão em apenas um dos conjuntos, excluindo aqueles que aparecem nos dois.
print(clientes_ativos ^ clientes_com_pedidos)
```

### 1.4 Dicionários — dict

```python
# Dicionários representam dados por chave e valor.
# Diferença entre [] e .get()
# pedido["categoria"] gera KeyError se a chave não existir.
# pedido.get("categoria") retorna None nesse caso, ou o valor padrão informado. Se não tiver "categoria", ele retorna "não informada"

pedido = {
    "id": 101,
    "cliente_id": 25,
    "valor": 150.50,
    "status": "concluido"
}

print(pedido["valor"])
print(pedido.get("categoria","não informada"))

print(pedido)

```

```python
# Percorrendo dicionarios 
pedido = {
    "id": 101,
    "cliente_id": 25,
    "valor": 150.50,
    "status": "concluido"
}

for chave, valor in pedido.items():
    print(f"{chave}: {valor}")
```

```python
# Dicionários aninhados
pedido = {"id": 101, "cliente": {"id": 25,"nome":"ana"},"itens":[{"produto":"notebook","valor": 3500},{"produto":"mouse","valor": 150}]}

print(pedido["cliente"]["nome"])
print(pedido["itens"][0]["produto"])
```

```python
# Exercício 3: Sets e clientes únicos

# União	A \| B	A.union(B)
# Interseção	A & B	A.intersection(B)
# Diferença	A - B	A.difference(B)
# Diferença simétrica	A ^ B	A.symmetric_difference(B)

clientes_cadastrados = [101, 102, 103, 104, 105, 105]
clientes_compradores = [102, 104, 106, 106]

# 1. Converta para sets

cadastrados = set(clientes_cadastrados)
compradores = set(clientes_compradores)
print(cadastrados)
print(compradores)

# 2. Interseção
print(cadastrados.intersection(compradores))
print(compradores & cadastrados)

# 3. União
print(cadastrados.union(compradores))
print(cadastrados | compradores)

# 4. Diferença
print(cadastrados.difference(compradores))
print(cadastrados - compradores)

# 5. Diferença simétrica
print(cadastrados.symmetric_difference(compradores))
print(cadastrados ^ compradores)

```
