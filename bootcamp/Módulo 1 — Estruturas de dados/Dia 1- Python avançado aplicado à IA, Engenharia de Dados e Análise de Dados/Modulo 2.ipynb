### 1. LIST COMPREHENSION

Uma list comprehension permite construir uma nova lista a partir de um iterável, aplicando transformações e filtros.

```python
pedidos = [
    {"id": 101, "status": "concluido", "valor": 450},
    {"id": 102, "status": "pendente", "valor": 200},
    {"id": 103, "status": "concluido", "valor": 350},
    {"id": 104, "status": "cancelado", "valor": 150}
]
```

Forma tradicional com for:

```python
pedidos_concluidos = []

for pedido in pedidos:
    if pedido["status"] == "concluido":
        pedidos_concluidos.append(pedido)

print(pedidos_concluidos)
```

Forma com list comprehension

[expressao for elemento in iteravel if condicao]

#expressao: o que será colocado na nova lista.
#elemento: variável que representa cada item.
#iteravel: coleção percorrida.
#condicao: filtro opcional.

```python
pedidos_concluidos = [
    pedido
    for pedido in pedidos
        if pedido["status"] == "concluido"
]

print(pedidos_concluidos)
```

Transformando valores - 
Também podemos extrair apenas os valores dos pedidos concluídos:

```python
pedidos = [
    {"id": 101, "status": "concluido", "valor": 450},
    {"id": 102, "status": "pendente", "valor": 200},
    {"id": 103, "status": "concluido", "valor": 350},
    {"id": 104, "status": "cancelado", "valor": 150}
]

valores = [
    pedido["valor"]
    for pedido in pedidos
        if pedido["status"] == "concluido"
]

print(valores)
```

### 2. DICT COMPREHENSION
Uma dict comprehension constrói um dicionário a partir de um iterável.

```python
# Imagine que precisamos consultar rapidamente um pedido por ID.

pedidos = [
    {"id": 101, "status": "concluido", "valor": 450},
    {"id": 102, "status": "pendente", "valor": 200},
    {"id": 103, "status": "concluido", "valor": 350},
    {"id": 104, "status": "cancelado", "valor": 150}
]

pedidos_por_id = {
    pedido["id"]: pedido
    for pedido in pedidos
}
print(pedidos_por_id[103])  

```

### 3. Iteradores: iter() e next()
Um iterável é um objeto que pode ser percorrido, como uma lista. Um iterador é um objeto que mantém o estado da iteração e entrega o próximo elemento quando solicitado.

```python
pedidos_ids = [101, 102, 103]

iterador = iter(pedidos_ids)
print(next(iterador))  # Saída: 101
print(next(iterador))  # Saída: 102
print(next(iterador))  # Saída: 103
```

### 4. Generators e yield
Um generator produz valores sob demanda, sem precisar construir uma lista com todos os resultados.

```python
def gerar_ids():
    return [1,2,3,4,5]
```

Exemplo com yield:

```python
def gerar_ids():
    for id_pedido in range(1,6):
        yield id_pedido

gerador = gerar_ids()

print(next(gerador))  # Saída: 1
print(next(gerador))  # Saída: 2
print(next(gerador))  # Saída: 3
```

### 5. Generator de lotes — Batch Processing
Imagine uma API que retorna milhares de registros. Em vez de processá-los todos de uma vez, podemos trabalhar com lotes menores.

```python
def gerar_lotes(registros, tamanho_lote):
    if tamanho_lote <= 0:
        raise ValueError("O tamanho do lote deve ser positivo")

    for inicio in range(0, len(registros), tamanho_lote):
        fim = inicio + tamanho_lote
        yield registros[inicio:fim]

registros = list(range(1,1001))

for lote in gerar_lotes(registros,100):
    print(
        f"Processando {len(lote)} registros "
        f" do ID {lote[0]} ao {lote[-1]}"
    )   


```

### Exercício 8 — Generators e yield
Agora vamos praticar funções geradoras, que permitem produzir valores sob demanda, sem construir uma lista completa de resultados.

```python
def gerar_ids():
    for id_pedido in range(1,6):
        yield id_pedido

gerador = gerar_ids()

print(next(gerador))  # Saída: 1
print(next(gerador))  # Saída: 2
print(next(gerador))  # Saída: 3
```

```python
# Sua tarefa
# Crie uma função chamada filtrar_valores(pedidos).
# Utilize for para percorrer os pedidos.
# Utilize if para selecionar apenas pedidos cujo valor seja maior que 200.
# Utilize yield para produzir somente o valor de cada pedido aprovado.
# Consuma o gerador utilizando outro for e imprima os valores.

pedidos = [
    {"id": 101, "valor": 450},
    {"id": 102, "valor": 200},
    {"id": 103, "valor": 350},
    {"id": 104, "valor": 100}
]

def filtrar_valores(pedidos):
    for pedido in pedidos:
        if pedido["valor"] > 200:
            yield pedido["valor"]

for valor in filtrar_valores(pedidos):
    print(valor)
```

```python
gerador = filtrar_valores(pedidos)
print(next(gerador))
print(next(gerador))
```

Imagine que você precise processar 1 milhão de registros provenientes de uma API, banco de dados ou arquivo.

Em vez de enviar todos os registros de uma vez para um modelo de IA ou banco de dados, você pode separá-los em lotes de 100, 500 ou 1.000 registros.

```python

from itertools import islice

def gerar_lotes(registros, tamanho_lote):
    if tamanho_lote <= 0:
        raise ValueError("O tamanho do lote deve ser maior que zero.")

    iterador = iter(registros)

    while True:
        lote = list(islice(iterador, tamanho_lote))

        if not lote:
            return

        yield lote


def gerar_registros():
    for id_registro in range(1, 1000_001):
        yield id_registro


registros = gerar_registros()

for lote in gerar_lotes(registros, 1000):
    print(
        f"Processando {len(lote)} registros "
        f"do ID {lote[0]} ao {lote[-1]}"
    )

```
