### Módulo 3: Funções, *args e **kwargs

```python
# Calcular subtotal 1 forma

def calcular_subtotal(quantidade, preco_unitario):
    subtotal = quantidade * preco_unitario
    return subtotal

resultado = calcular_subtotal(5, 10.0)
print(f"O subtotal é: {resultado}")
```

### 2. O que é *args? 
O *args permite receber uma quantidade variável de argumentos posicionais. Armazena uma tuple

```python
def somar_valores(*args):
    print(args)
    return sum(args)

resultado_soma = somar_valores(1, 2, 3, 4, 5)
print(f"A soma dos valores é: {resultado_soma}")
```

### 3. O que é **kwargs?
O **kwargs permite receber uma quantidade variável de argumentos nomeados. Armazena uma dict

```python
def registrar_evento(**kwargs):
    print(kwargs)

registrar_evento(
    evento = "pipeline_atualizado",
    status = "sucesso",
    registros=1500
)
```

Combinando *args e **kwargs

```python
def registrar_metricas(*valores, **metadados):
    print("valores:",valores)
    print("Metadados:", metadados)

registrar_metricas(
    100, 200, 300,
    origem="api",
    ambiente="producao"
)
```

### 4. Diferença entre print, return e yield

print()- Exibe uma informação, normalmente no terminal
return - Encerra a função e devolve um resultado
yield - Produz um valor e suspende a execução da função geradora

Exercício 10 — Calcular faturamento
Você recebeu os pedidos abaixo e precisa desenvolver uma função para calcular o faturamento dos pedidos concluídos.

```python
pedidos = [
    {"id": 101, "status": "concluido", "valor": 450},
    {"id": 102, "status": "pendente", "valor": 200},
    {"id": 103, "status": "concluido", "valor": 350},
    {"id": 104, "status": "cancelado", "valor": 150},
    {"id": 105, "status": "concluido", "valor": 500}
]

def calcular_faturamento(pedidos):
    faturamento = 0

    for pedido in pedidos:
        
        if pedido["status"] == "concluido":
            faturamento += pedido["valor"]

    return faturamento
    
resultado = calcular_faturamento(pedidos)
print(f"Faturamento: R${resultado:.2f}") 
```

Exercício 11 — Encontrar o maior valor - Você está desenvolvendo uma função para identificar o maior tempo de processamento de uma série de requisições a uma API de IA.

A quantidade de requisições varia, portanto a função deve aceitar um número variável de argumentos.


```python
def maior_valor(*args):
    if not args:
        return None
    return max(args)

print(maior_valor(10, 50, 30, 80, 20))
print(maior_valor(150, 200))
print(maior_valor(7))
print(maior_valor())
```

Exercício 12 — Registrar metadados de execução

```python
# Imagine um pipeline que ingere registros de uma API e os grava no Snowflake ou Databricks.
# Você precisa criar uma função que registre informações sobre cada execução. Entretanto, nem todas as execuções possuem os mesmos metadados.
# Uma execução pode conter apenas status e registros. Outra pode incluir duracao, origem, ambiente ou erro.


def registrar_execucao(pipeline,**kwargs):
    log = {
        **kwargs,
        "pipeline": pipeline
    }

    return log

registrar_execucao(
    "ingestao_clientes",
    status="sucesso",
    registros=1500, 
    duracao=12.2
)

registrar_execucao(
    "geracao_embeddings",
    status="erro",
    erro="Tempo limite excedido",
    tentativas=3
)

```

### Aula 1 — Módulo 3: Exercício 13 — print(), return e yield
Python intermediário · Engenharia de IA e Dados
Você concluiu os exercícios 10, 11 e 12 com nota 10,0. Agora vamos encerrar o Módulo 3 com um conceito fundamental: a diferença entre exibir um valor, retornar um resultado e produzir valores progressivamente.

A. print() — Exibir

```python
def calcular_total():
    total = 100+200
    print(total)

resultado = calcular_total()
print(resultado)
```

B. return — Devolver

```python
def calcular_total():
    total = 100 + 200
    return total

resultado = calcular_total()
print(resultado)
```

C. yield — Produzir valores progressivamente

```python
def gerar_ids():
    yield 101
    yield 102
    yield 103

resultado = gerar_ids()

print(next(resultado))
print(next(resultado))
print(next(resultado))
```
